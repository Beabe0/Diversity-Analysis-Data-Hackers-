```python
from scipy.stats import chi2_contingency #compara os resultados da tabela com como se as duas variaveis fossem independentes. Quanto maior a diferença entre a distribuição observada da esperada, maior será o valor, o que indica uma forte assosiação entre as variáveis
```

```python
def cramer_coeficiente(coluna1, coluna2):
  tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
  chi2 = chi2_contingency(tabela_cruzada)[0] #essa função retorna muitos valores. mas como só queremos o qui quadrado, que é o primeiro, colocamos 0 entre[]
  soma = np.sum(tabela_cruzada)
  mini = min(tabela_cruzada.shape) -1
  cramer = np.sqrt(chi2/(soma*mini))
  return cramer
```

```python
cramer_coeficiente(dados['COR/RACA/ETNIA'], dados['NIVEL DE ENSINO']) #resultado: nao tem quase nenhuma correlação
```

```python
cramer_coeficiente(dados['GENERO'], dados['NIVEL DE ENSINO']) #resultado: tem um pouco mais de correlação, mas pouco
```

```python
cramer_coeficiente(dados['GENERO'], dados['SALARIO']) #correlacao alta
```

```python
cramer_coeficiente(dados['COR/RACA/ETNIA'], dados['SALARIO']) #correlacao alta
```
